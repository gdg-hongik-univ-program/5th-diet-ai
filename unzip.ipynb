```python
from google.colab import drive
drive.mount('/content/drive')
```

```python
%%bash
set +e
# 대상 폴더 경로 (본인의 경로에 맞게 수정하세요)
BASE_PATH="/content/drive/MyDrive/법률 규정 텍스트 분석 데이터 (고도화)- 상황에 따른 판례 데이터(Sublabel)"

find "$BASE_PATH" -name "*.zip" | while read z; do
  # 1. 파일이 들어있는 디렉토리 경로 추출 (예: /content/Training/cat)
  d=$(dirname "$z")

  # 2. 파일명에서 확장자 제거 (예: data.zip -> data)
  folder_name=$(basename "$z" .zip)

  # 3. 새로운 목적지 경로 생성 (예: /content/Training/cat/data)
  target_dir="$d/$folder_name"

  echo "Unzipping $z into $target_dir"

  # 4. 압축 해제 (-d 옵션이 폴더를 자동으로 생성해줍니다)
  unzip -O cp949 -o -qq "$z" -d "$target_dir" || echo "Failed: $z"
done
```

```python
%%bash
# 대상 디렉토리 설정 (본인의 경로로 수정하세요)
TARGET_DIR="/content/drive/MyDrive/행정법 LLM 사전학습 및 Instruction Tuning 데이터(Training)/02.라벨링데이터"

echo "--- 폴더별 파일 개수 리포트 ---"

# 해당 디렉토리 내의 모든 폴더(디렉토리)를 찾아 반복문 실행
for d in "$TARGET_DIR"/*/; do
    # 폴더 이름만 추출
    folder_name=$(basename "$d")

    # 해당 폴더 내의 모든 파일(-type f) 개수 세기 (하위 폴더 포함)
    count=$(find "$d" -type f | wc -l)

    echo "[$folder_name]: $count 개"
done

echo "------------------------------"
```
